```python
# TASK 5 :
today_orders = pd.DataFrame({
    "order_id": [1, 2, 3],
    "item": ["Pizza", "Burger", "Pasta"],
    "price": [250, 150, 200]
})

yesterday_orders = pd.DataFrame({
    "order_id": [4, 5, 6],
    "item": ["Sandwich", "Fries", "Coffee"],
    "price": [100, 80, 120]
})

df = pd.concat([today_orders, yesterday_orders])

df = df.reset_index(drop=True)
df
print(df)
```
